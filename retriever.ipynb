```python
from neo4j_module.connection import Neo4jConnection
neo4j_con = Neo4jConnection()
driver = neo4j_con.get_driver()
```

```python
from embedding.query_embedder import embed
from embedding.vector_search import perform_search

query = "What issues are reported by Tsunoda San"
vector = embed(query)
with driver.session() as session:
    graph_result = perform_search(session, vector, "ticket_embedding_index", 100)
```

```python
print(len(graph_result))
for result in graph_result[:200]:
    print(result['node']['issue_key'], result['node']['description_en'], result['score']) 
```
