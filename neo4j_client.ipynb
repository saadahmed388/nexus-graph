```python
import os
import json
from dotenv import load_dotenv
from pathlib import Path
import tiktoken
load_dotenv()
```

```python
def read_json(filepath):
    filepath = Path(filepath)
    if filepath.is_file():
        try:
            with open(filepath, "r", encoding='utf-8') as f:
                return json.load(f)
        except json.JSONDecodeError:
            return []
    return []

filepath = "./data/json-exports/schema_v4.json"
incidents = read_json(filepath)
print(len(incidents))
```

```python
from neo4j_module.connection import Neo4jConnection
from neo4j_module.constraints import create_constraints
from neo4j_module.importer import import_ticket
neo4j_con = Neo4jConnection()
driver = neo4j_con.get_driver()

with driver.session() as session:
    for incident in incidents:
        session.execute_write(
            import_ticket,
            incident
        )

print("Tickets Imported")



```
