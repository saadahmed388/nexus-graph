```python
import os
import json
from dotenv import load_dotenv
from pathlib import Path
import tiktoken
load_dotenv()
```

```python
class Neo4jClient:
    def __init__(self):
        self.user = os.getenv("NEO4J_USER")
        self.password = os.getenv("NEO4J_PASS")
        self.key = os.getenv("NEO4J_URI")
        self.load_json()

    def load_json(self, filepath = "./json-exports/schema_v1.json"):
        filepath = Path(filepath)
        if filepath:
            with open(filepath, "r", encoding='utf-8') as f:
                self.inc_list = json.load(f)

    def export_bug_comments(self, outfilepath = "./json-exports/bug_comments_v1.json"):
        outfilepath = Path(outfilepath)
        c_list = []

        for inc in self.inc_list:
            c_dict = {}
            if inc["identity"]["issue_type"] == "Bug":
                c_dict["issue_key"] = inc["identity"]["issue_key"]
                c_dict["comments"] = inc["comments"]
                c_list.append(c_dict)

        with open(outfilepath, "w", encoding='utf-8') as f:
            json.dump(c_list, f, indent=4)

    def count_tokens(self, filepath = "./json-exports/bug_comments_v1.json"):

        with open("incidents.json", "r", encoding="utf-8") as f:
            data = json.load(f)
        enc = tiktoken.get_encoding("o200k_base")  # Suitable for modern GPT-4.1/GPT-5 family
        json_text = json.dumps(data, ensure_ascii=False)
        num_tokens = len(enc.encode(json_text))
        print(f"Total tokens: {num_tokens:,}")



            
        
                

```

```python
neo_client = Neo4jClient()
neo_client.load_json()
neo_client.export_bug_comments()
inc_list = neo_client.inc_list
print(json.dumps(inc_list[0], indent=4))
```

```python
persons_set = set()
labels_set = set()
tracks_set = set()
systems_set = set()
environments_set = set()
sprints_set = set() #Not Included
departments_set = set() #Not Included
issue_type_set = set()
bug_category_set = set()

for inc in inc_list:
    for w in inc["people"]["watchers"]:
        persons_set.add(w)
    for l in inc["classification"]["labels"]:
        labels_set.add(l)
    for e in inc["system_context"]["impacted_environments"]:
        environments_set.add(e)

    tracks_set.add(inc["classification"]["track"])
    systems_set.add(inc["system_context"]["force_system"])
    sprints_set.add(inc["planning"]["sprint"])
    departments_set.add(inc["organizational_context"]["department"])
    issue_type_set.add(inc["Issue Type"])
    bug_category_set.add(inc[""])


print("Persons: ", persons_set)
print("Labelsl: ", labels_set)
print("Tracks: ", tracks_set)
print("Systems: ", systems_set)
print("Environements", environments_set)
print("Sprints: ", sprints_set)
print("Departments: ", departments_set) 
```
