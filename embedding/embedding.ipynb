```python
import json

infile_path = r"./doc-file/vector_doc_v1_embedded.json"
with open(infile_path, "r", encoding="utf-8") as f:
    print("loading...")
    data = json.load(f)


outfile_path = r"./doc-file/vector_doc_v1_embedded_compressed.jsonl"
with open(outfile_path, "w", encoding="utf-8") as f_out:
    print("saving...")
    for item in data:
        # json.dumps keeps your nested lists on a single line safely
        f_out.write(json.dumps(item, ensure_ascii=False) + "\n")
```
